# CRISP-DM Phase 1: Business Understanding
## 1. Current Market Background (January 2026)
As of late January, the Mag 7 are no longer moving in lockstep. This "Great Divergence" is the defining characteristic of the current market. 
- **The Performance Gap**
    - Alphabet (GOOGL) & Amazon (AMZN): Currently the leaders of the pack. Alphabet is riding high on Gemini AI integration, while Amazon is seeing massive ROI from its cloud (AWS) automation.
    - Apple (AAPL) & Tesla (TSLA): Currently in a technical downtrend. Apple is struggling with fewer near-term catalysts, and Tesla is facing valuation pressure as investors demand proof of robotaxi profitability.
- **Macro Shift**: There is a notable "rotation" occurring. Investors are moving capital out of high-valuation tech and into "Value" stocks and the "S&P 493" (the rest of the market).

## 2. Business Objectives
- **Primary Goal:** Predict the directional movement (Binary: 1 for Gain, 0 for Loss/Flat) of individual Mag 7 stocks for the next trading day
- **Target Accuracy:** At least 53% correct prediction

## 3. Risks and Contingencies
- **The "Earnings Gap" Risk:** Technical indicators often fail during earnings weeks because the price movement is driven by news, not past price patterns.
    - *Contigency*: Exclude earnings days from the training set or add a "Days Until Earnings" feature to the model.
- **Data Latency:** Free tier APIs sometimes have 15-minutes delay
    - *Contigency*: Always use the Adj Close column to account for dividends and splits, which are common in these high-value stocks.
- **Overfitting to high growth stocks**: High growth stocks like Nvidia can cause overfitting during model training and the model may fail to predict more stagnant stocks like Apple
    - *Contigency*: Train individual models for each stock rather than one "Master Mag 7" model.

## 4. Assumption and Constraints
### Assumption
- Historical price patterns are assumed to have predictive power for future movements; we assume no "black swan"(severe, unpredicable) events occur during the testing period (like recession).
- No presence of catalyst news events which may drastically affect prices. Movement of stocks prices comes purely from technical analysis.

### Constraints
- Limited to 5 years of historical data to allow usage of standard laptop without any external TPU or GPU needed.
- No high-frequency trading capabilities.
