```python
import yfinance as yf
import pandas as pd
import numpy as np
```

Pulling 5 years of open and close prices of the Magnificent 7 Stocks in US Stock Market.

```python
tickers = ["AAPL", "MSFT", "GOOGL", "AMZN", "META", "NVDA", "TSLA"]
start_date = "2021-01-01"
end_date = "2026-01-01"
open_close_data = yf.download(tickers, start=start_date, end=end_date)
open_close_data
```

```python
pd.DataFrame({
    'missing' : open_close_data.isna().sum()
}).sort_values('missing', ascending=False)
```

There are no missing value in the data. However, pulling all 7 tickers info cause the data to contain too many columns. Reducing the dataset by grouping it by ticker may make future data processing much efficient.

```python
#Get individual ticker open, close, low, high, volume data
appl_data = yf.download("AAPL", start_date, end_date).reset_index()
msft_data = yf.download("MSFT", start_date, end_date).reset_index()
googl_data = yf.download("GOOGL", start_date, end_date).reset_index()
amzn_data = yf.download("AMZN", start_date, end_date).reset_index()
meta_data = yf.download("META", start_date, end_date).reset_index()
nvda_data = yf.download("NVDA", start_date, end_date).reset_index()
tsla_data = yf.download("TSLA", start_date, end_date).reset_index()

#Save data into csv file for processing (keep index True to get date column)
appl_data.to_csv("appl.csv", index=True)
msft_data.to_csv("msft.csv", index=True)
googl_data.to_csv("googl.csv", index=True)
amzn_data.to_csv("amzn.csv", index=True)
meta_data.to_csv("meta.csv", index=True)
nvda_data.to_csv("nvda.csv", index=True)
tsla_data.to_csv("tsla.csv", index=True)
```
