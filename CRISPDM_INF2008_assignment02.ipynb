## Structural Cleaning Before Pipeline Construction

A small number of preprocessing steps are applied **before** the train–test split because they are **deterministic structural corrections**, not learned transformations.

These include:

- removing known leakage columns
- dropping structurally unsuitable columns
- standardising an ambiguous category label
- removing logically invalid records

These actions do **not** use dataset statistics such as mean, median, mode, nearest neighbours, or target-based relationships, so they do not introduce data leakage.

````python
df = df_raw.copy()

TARGET_COL = "is_canceled"

LEAKAGE_COLS = [
    "reservation_status",
    "reservation_status_date"
]

DROP_COLS = [
    "company"
]

# Remove leakage columns and structurally unsuitable columns
df = df.drop(columns=LEAKAGE_COLS + DROP_COLS, errors="ignore")

# Standardise ambiguous meal label
df["meal"] = df["meal"].replace("Undefined", "None")

# Remove logically invalid bookings:
# no adults, no children, and no babies
before_rows = len(df)
df = df[~((df["adults"] == 0) & (df["children"].fillna(0) == 0) & (df["babies"] == 0))]
after_rows = len(df)

print(f"Rows removed due to invalid booking composition: {before_rows - after_rows}")
print("Shape after structural cleaning:", df.shape)
df.head()
````

## Feature–Target Separation

The target variable is **`is_canceled`**, where:

- `1` = booking was cancelled
- `0` = booking was not cancelled

At this stage, predictors and target are separated before any modelling workflow begins.

````python
y = df[TARGET_COL]
X = df.drop(columns=[TARGET_COL])

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution (overall):")
print(y.value_counts())
print("\nTarget distribution (overall proportion):")
print(y.value_counts(normalize=True).rename("proportion"))
````

## Train–Test Split

The dataset is divided into **training** and **test** sets using an **80/20 split** with stratification on the target variable.

This split is performed **before** feature engineering, imputation, encoding, scaling, and SMOTE so that all learned preprocessing steps remain isolated from the test set.

````python
nan_indices = y[y.isna()].index
X_cleaned = X.drop(nan_indices)
y_cleaned = y.drop(nan_indices)

X_train, X_test, y_train, y_test = train_test_split(
    X_cleaned,
    y_cleaned,
    test_size=0.20,
    stratify=y_cleaned,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("\nTarget distribution (train):")
print(y_train.value_counts(normalize=True))

print("\nTarget distribution (test):")
print(y_test.value_counts(normalize=True))
````

## Leakage-Safe Feature Engineering

Feature engineering is applied **after the train–test split** to keep the modelling workflow explicit and leakage-safe.

The engineered features below are **deterministic domain-based features** derived only from existing booking attributes:

- `total_nights` = weekend nights + week nights
- `total_guests` = adults + children + babies
- `is_family` = whether children or babies are included
- `has_agent` = if travel agency booked the room

These features do not use target information or dataset-level statistics.

````python
def add_booking_features(df_in):
    df_out = df_in.copy()

    df_out["children"] = df_out["children"].fillna(0)

    df_out["total_nights"] = (
        df_out["stays_in_weekend_nights"] + df_out["stays_in_week_nights"]
    )

    df_out["total_guests"] = (
        df_out["adults"] + df_out["children"] + df_out["babies"]
    )

    df_out["is_family"] = (
        ((df_out["children"] > 0) | (df_out["babies"] > 0)).astype(int)
    )
    df_out["has_agent"] = df_out['agent'].notna().astype(int)

    return df_out


X_train = add_booking_features(X_train)
X_test = add_booking_features(X_test)

print("Feature engineering applied.")
print("X_train shape after feature engineering:", X_train.shape)
print("X_test shape after feature engineering:", X_test.shape)
````

## Feature Grouping for Pipeline Engineering

The features are divided into:

## Preprocessing Design

The preprocessing pipeline applies different transformations to numeric and categorical features.

### Binary branch
- most-frequent imputation for missing labels
- use variance threshold to drop near constant columns (99% zeroes)

### Numeric branch
- median imputation for missing values
- robust scaling to reduce the effect of outliers

### Categorical branch
- most-frequent imputation for missing labels
- one-hot encoding for machine-readable representation

### Target Encoding for `agent` column
Among the set of features, `agent` column is a nominal categorical variable which is the identifier of the travel agency that made the booking. Doing classical numerical imputation(mean, median, mode) for the missing values would not provide much value. Missing values may not be random, which will provide meaningful signal for the models to capture.

**Target encoding** can be used to replace category values with the mean of the target value. For example:
```
agent_id 42  → appears 20 times → 4 of those were cancellations  → encoded as 0.20
```
The above is the naive version which leaks the target in the training data. This is especially bad for rare agents (e.g. an agent with only 1 booking that was cancelled → encoded as 1.0, a perfect but meaningless signal).

To fix this, smoothing is done to blend category mean and global mean based on the number of samples.

$$\text{encoded} = \frac{n \cdot \text{category\_mean} + m \cdot \text{global\_mean}}{n + m}$$
```
where
n = number of rows for that category
m = smoothing factor (hyperparameter, default ~1)
```
Rare agents get pulled heavily toward the global mean. Common agents stay close to their true mean.

Target Encoding is done **within** the pipeline to fit on each training fold independently during the Cross-validation.

After preprocessing, **SMOTE** is applied to the training data within the pipeline to address class imbalance.  
Finally, **Logistic Regression** is used as the baseline classifier.

Because all learned steps are placed inside the pipeline, the transformations are fitted only on the training workflow, preventing data leakage.

````python
# Binary feature type preprocessing
binary_transformer = SkPipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("variance_filter", VarianceThreshold(threshold=0.01))
])

# Continuous preprocessing
continuous_transformer = SkPipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", RobustScaler())
])

# Categorical preprocessing
categorical_transformer = SkPipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

agent_transformer = SkPipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("target_enc", TargetEncoder(smooth="auto"))
])

# Combine preprocessing branches
preprocessor = ColumnTransformer(
    transformers=[
        ("binary", binary_transformer, binary_features),
        ("num", continuous_transformer, continuous_features),
        ("cat", categorical_transformer, categorical_features),
        ("agent", agent_transformer, ["agent"]),
    ],
    remainder="drop"
)

# Leakage-safe Stage 2 pipeline
stage2_pipeline = ImbPipeline(steps=[
    ("preprocessor", preprocessor),
    ("smote", SMOTE(random_state=42)),
    ("classifier", LogisticRegression(max_iter=2000, random_state=42))
])

stage2_pipeline
````

# Part B: Champion Model Selection

# Candidate Model Pipeline

Each model is wrapped inside the same **ImbPipeline** structure from Part A. The pipeline runs three steps in sequence for every fold of cross-validation:

1. **Preprocessor**: applies imputation, encoding, and scaling to the raw features
2. **SMOTE**: generates synthetic minority class samples to address class imbalance, but only on the training fold so that validation data is never contaminated
3. **Classifier**: the model that learns from the processed training data and makes predictions

Keeping all three steps inside one pipeline prevents data leakage and ensures every model is evaluated under identical conditions.

However, the default Logistic Regression tends to perform the worst in imbalanced datasets because it treats all classes equally and its predicted probabilities cluster near 0.5. Two targeted adjustments are applied to give it a fair chance:

- `class_weight='balanced'` — puts more weight on cancellations during training so the model does not ignore the minority class
- **Threshold lowered to 0.35** — the hotel bookings dataset has approximately 37% cancellations, so a 35% threshold matches the natural rate of cancellations in the data. This means the model flags a booking as a cancellation if its predicted probability is at least 35%, instead of the default 50%. This helps Logistic Regression catch more cancellations that it would otherwise miss.

Random Forest and Hist Gradient Boosting are evaluated at the default 50% threshold as they already capture feature interactions well and do not need this correction.

````python
# Build the 3 candidate pipelines using the same preprocessor from Part A
preprocessor.set_params(sparse_threshold=0)
candidate_pipelines = {

    "Logistic Regression": ImbPipeline(steps=[
        ("preprocessor", preprocessor),
        ("smote",        SMOTE(random_state=42)),
        ("classifier",   LogisticRegression(
                            C=1.0,
                            class_weight="balanced",
                            max_iter=2000,
                            random_state=42))
    ]),

    "Random Forest": ImbPipeline(steps=[
        ("preprocessor", preprocessor),
        ("smote",        SMOTE(random_state=42)),
        ("classifier",   RandomForestClassifier(
                             n_estimators=200,
                             min_samples_leaf=5,
                             n_jobs=-1,
                             random_state=42))
    ]),

    "Hist Gradient Boosting": ImbPipeline(steps=[
        ("preprocessor", preprocessor),
        ("smote",        SMOTE(random_state=42)),
        ("classifier",   HistGradientBoostingClassifier(
                             max_iter=200,
                             learning_rate=0.05,
                             max_leaf_nodes=31,
                             random_state=42))
    ]),

}
````

# 5-Fold Stratified Cross-Validation

**How it works:**


The training data is divided into 5 equal parts called folds. In each round, 4 folds are used to train the model and the remaining 1 fold is used to test it. This process repeats 5 times, rotating which fold is used for testing each time. The final reported score is the average across all 5 rounds, along with the standard deviation to show how consistent the model was.

**How it supports fair model selection:**


Using 5 folds instead of a single train/test split means the model is tested on 5 different slices of data. This reduces the risk that a lucky or unlucky split skews the result. The standard deviation also reveals whether a model performs consistently or whether its score swings widely depending on what data it sees — a stable model is more trustworthy in production.


**Stratified** means each fold is constructed so that the proportion of cancelled vs not-cancelled bookings stays the same as in the full dataset. This prevents a fold from accidentally containing too many or too few cancellations, which would distort the evaluation.


````python
# Set up 5-fold stratified cross-validation
# Stratified means each fold keeps the same class ratio as the full dataset
cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Define the metrics to collect across all folds
scoring_default = {
    "f1":        make_scorer(f1_score, zero_division=0),
    "recall":    "recall",
    "precision": "precision",
    "roc_auc":   "roc_auc",
}

# Custom scorers for Logistic Regression using 0.35 threshold
# 0.35 is chosen because the dataset has ~37% cancellations — threshold matches the natural rate
def f1_at_35(clf, X, y):
    proba = clf.predict_proba(X)[:, 1]
    pred  = (proba >= 0.35).astype(int)
    return f1_score(y, pred, zero_division=0)

def recall_at_35(clf, X, y):
    proba = clf.predict_proba(X)[:, 1]
    pred  = (proba >= 0.35).astype(int)
    return recall_score(y, pred, zero_division=0)

def precision_at_35(clf, X, y):
    proba = clf.predict_proba(X)[:, 1]
    pred  = (proba >= 0.35).astype(int)
    return precision_score(y, pred, zero_division=0)

def roc_auc_scorer(clf, X, y):
    proba = clf.predict_proba(X)[:, 1]
    return roc_auc_score(y, proba)

scoring_lr = {
    "f1":        f1_at_35,
    "recall":    recall_at_35,
    "precision": precision_at_35,
    "roc_auc":   roc_auc_scorer,
}

# Run cross-validation for each model and store the results
cv_results = {}

for name, pipe in candidate_pipelines.items():
    print(f"Running cross-validation for: {name} ...")
    # Use threshold-adjusted scoring only for Logistic Regression
    active_scoring = scoring_lr if name == "Logistic Regression" else scoring_default
    scores = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=cv_strategy,
        scoring=active_scoring,
        return_train_score=False,
    )
    cv_results[name] = scores
    print(f"  F1 Score : {scores['test_f1'].mean():.4f} ± {scores['test_f1'].std():.4f}")
    print(f"  Recall   : {scores['test_recall'].mean():.4f} ± {scores['test_recall'].std():.4f}")
    print(f"  Precision: {scores['test_precision'].mean():.4f} ± {scores['test_precision'].std():.4f}")
    print(f"  ROC-AUC  : {scores['test_roc_auc'].mean():.4f} ± {scores['test_roc_auc'].std():.4f}")
    print()
````

## Cross-Validation Results Summary Table

The summary table below presents the **mean and standard deviation** of each metric across all 5 folds.

- **Mean** reflects the average performance of the model across 5 different test slices
- **Std** reflects how stable that performance is — a lower std means the model generalises consistently regardless of which data it is tested on

The table is sorted by **F1 Score (mean)** from highest to lowest, as F1 is the primary metric used for Champion selection.

````python
rows = []
for name, scores in cv_results.items():
    rows.append({
        "Model"            : name,
        "F1 (mean)"        : round(scores["test_f1"].mean(),        4),
        "F1 (std)"         : round(scores["test_f1"].std(),         4),
        "Recall (mean)"    : round(scores["test_recall"].mean(),    4),
        "Recall (std)"     : round(scores["test_recall"].std(),     4),
        "Precision (mean)" : round(scores["test_precision"].mean(), 4),
        "Precision (std)"  : round(scores["test_precision"].std(),  4),
        "ROC-AUC (mean)"   : round(scores["test_roc_auc"].mean(),   4),
        "ROC-AUC (std)"    : round(scores["test_roc_auc"].std(),    4),
    })

summary_df = (pd.DataFrame(rows)
                .sort_values("F1 (mean)", ascending=False)
                .reset_index(drop=True))

display(summary_df)
````

## Controlled Ablation Setup – Random Forest Champion Model

This section performs a **controlled ablation study** on the selected **Champion Random Forest** model. The goal is to test whether small, isolated changes to the model configuration improve generalization performance.

**Step 1: Understand the Champion Model**  
The fitted Random Forest champion is first inspected to understand how it behaves internally. This includes examining fitted tree depth, leaf-level statistics, and feature importance. These observations are used as evidence for selecting which hyperparameters and feature engineering ideas are worth testing.

**Step 2: Design Experiments Based on Inspection Evidence**  
Using the findings from Step 1, four controlled experiments are selected. Each experiment is designed to test one specific hypothesis about either model tuning or feature representation, while keeping the rest of the pipeline fixed.

**Step 3: Run Controlled Experiments**  
Each experiment is executed by changing only the targeted parameter or feature configuration, with all other preprocessing, resampling, model settings, and cross-validation strategy held constant. Performance is evaluated using 5-fold cross-validation with F1-score as the primary metric, supported by Precision, Recall, and ROC-AUC.

**Step 4: Record Results in the Ablation Log**  
The outcome of each experiment is summarised in an Ablation Log table. For each case, the table records the hypothesis, controlled change, cross-validation metric impact, and conclusion, so that results are directly comparable across experiments.

**Step 5: Summarise Findings**  
The final step consolidates the results across all experiments. The aim is to determine whether each hypothesis is supported, identify which changes improve or weaken the champion model, and highlight any trade-offs between predictive performance and model complexity.

## Experiment 1: Hyperparameter Tuning - `max_depth`

### Hypothesis
The champion's trees grow to an **average fitted depth of 47.4** with no constraint.
We hypothesise that constraining `max_depth` to low values will cause a **large, measurable F1
degradation** because hotel cancellation relies on deep multi-feature interaction chains
(e.g., long lead time + Online TA + No Deposit + zero special requests) that shallow trees
cannot disentangle in a single path.

Conversely, if the best depth is `None`, it confirms the original champion setting was already
at the performance ceiling for this parameter.

### Controlled change
`max_depth` ∈ {5, 10, 15, 20, 30, None} - all other parameters and the full pipeline held constant.
Each pipeline is produced via `clone(baseline_pipe_C)` to guarantee no shared state between runs.

````python
max_depth_candidates = [5, 10, 15, 20, 30, None]

exp1_rows = []
print("Experiment 1 - systematic max_depth search  (all other params fixed)")
print("-" * 62)

for max_d in max_depth_candidates:
    pipe = clone(baseline_pipe_C)
    pipe.set_params(classifier__max_depth=max_d)
    cv_s = cross_validate(
        pipe, X_train, y_train,
        cv=cv_strategy, scoring=scoring_partC, n_jobs=-1
    )
    row = {
        "max_depth":       max_d,
        "label":           str(max_d) if max_d is not None else "None",
        "F1 Mean":         np.mean(cv_s["test_f1"]),
        "F1 Std":          np.std(cv_s["test_f1"]),
        "Recall Mean":     np.mean(cv_s["test_recall"]),
        "Precision Mean":  np.mean(cv_s["test_precision"]),
        "Precision Std":   np.std(cv_s["test_precision"]),
        "ROC-AUC Mean":    np.mean(cv_s["test_roc_auc"]),
        "ROC-AUC Std":     np.std(cv_s["test_roc_auc"]),
    }
    exp1_rows.append(row)
    tag = f"{str(max_d):<5s}" if max_d is not None else "None "
    print(f"  max_depth={tag}  -> F1: {row['F1 Mean']:.4f} +/- {row['F1 Std']:.4f}  |  "
          f"Prec: {row['Precision Mean']:.4f}")

exp1_df = pd.DataFrame(exp1_rows)

best_idx  = exp1_df["F1 Mean"].idxmax()
worst_idx = exp1_df["F1 Mean"].idxmin()
best_row  = exp1_df.loc[best_idx]
worst_row = exp1_df.loc[worst_idx]
delta_pp  = (best_row["F1 Mean"] - worst_row["F1 Mean"]) * 100

print(f"\n  Best  config : max_depth={best_row['label']}  -> F1 = {best_row['F1 Mean']:.4f}")
print(f"  Worst config : max_depth={worst_row['label']}  -> F1 = {worst_row['F1 Mean']:.4f}")
print(f"  F1 range across search       : {delta_pp:.2f} pp")
print(f"  Change vs baseline (best)    : {(best_row['F1 Mean']-b_f1_mean)*100:+.2f} pp")
````

## Experiment 3: Hyperparameter Tuning - `min_samples_leaf`
### Hypothesis
Model inspection revealed that the average number of **samples per leaf is 26.95**,
which is **5× the configured minimum** of `min_samples_leaf=5`.
This gap suggests that the current leaf constraint is not tightly binding - trees are
building larger leaves than required, so the parameter has meaningful regularisation headroom.

We hypothesise:
- **Very small values (e.g., 1):** Leaves may become too granular, overfitting to noise in individual folds, causing F1 to drop.
- **Values near 5–20:** Performance should remain stable near the baseline.
- **Very large values (e.g., 50–100):** Leaves become too coarse, losing granularity needed to distinguish borderline cases. Both F1 and Precision should degrade.

### Controlled change
`min_samples_leaf` ∈ {1, 5, 10, 20, 50, 100} - all other parameters held constant.
Each pipeline is produced via `clone(baseline_pipe_C)` to guarantee no shared state.

````python
min_leaf_candidates = [1, 5, 10, 20, 50, 100]

exp3_rows = []
print("Experiment 3 - systematic min_samples_leaf search  (all other params fixed)")
print("-" * 68)

for msl in min_leaf_candidates:
    pipe = clone(baseline_pipe_C)
    pipe.set_params(classifier__min_samples_leaf=msl)
    cv_s = cross_validate(
        pipe, X_train, y_train,
        cv=cv_strategy, scoring=scoring_partC, n_jobs=-1
    )
    row = {
        "min_samples_leaf": msl,
        "label":            str(msl),
        "F1 Mean":          np.mean(cv_s["test_f1"]),
        "F1 Std":           np.std(cv_s["test_f1"]),
        "Recall Mean":      np.mean(cv_s["test_recall"]),
        "Precision Mean":   np.mean(cv_s["test_precision"]),
        "Precision Std":    np.std(cv_s["test_precision"]),
        "ROC-AUC Mean":     np.mean(cv_s["test_roc_auc"]),
        "ROC-AUC Std":      np.std(cv_s["test_roc_auc"]),
    }
    exp3_rows.append(row)
    print(f"  min_samples_leaf={msl:<4d}  -> F1: {row['F1 Mean']:.4f} +/- {row['F1 Std']:.4f}  |  "
          f"Prec: {row['Precision Mean']:.4f}")

exp3_df = pd.DataFrame(exp3_rows)

e3_best_idx  = exp3_df["F1 Mean"].idxmax()
e3_worst_idx = exp3_df["F1 Mean"].idxmin()
e3_best_row  = exp3_df.loc[e3_best_idx]
e3_worst_row = exp3_df.loc[e3_worst_idx]
e3_delta_pp  = (e3_best_row["F1 Mean"] - e3_worst_row["F1 Mean"]) * 100

print(f"\n  Best  config : min_samples_leaf={e3_best_row['label']}  -> F1 = {e3_best_row['F1 Mean']:.4f}")
print(f"  Worst config : min_samples_leaf={e3_worst_row['label']}  -> F1 = {e3_worst_row['F1 Mean']:.4f}")
print(f"  F1 range across search           : {e3_delta_pp:.2f} pp")
print(f"  Change vs baseline (best config) : {(e3_best_row['F1 Mean'] - b_f1_mean)*100:+.2f} pp")
````

## Experiment 4: Feature Engineering – Country Region Encoding (3 Variants)

### Context

The feature `country_PRT` is the single highest-importance feature (Gini = 0.091), yet the current pipeline one-hot encodes `country` into **~169 sparse binary columns**, most of which contribute negligible importance. This suggests that predictive signal is concentrated in a small subset of countries (e.g., Portugal, UK, France, Spain, Germany), while the remaining categories introduce sparsity and noise.

To address this, a **region encoding approach** is introduced. Each ISO-3 country code is mapped to one of 8 broader geographic regions, and the resulting `region` feature is one-hot encoded. This reduces dimensionality (from ~169 to 8 categories) while aiming to preserve meaningful travel behaviour patterns that may cluster at a regional level.

| Code | Region |
|---|---|
| WE | Western Europe |
| EE | Eastern Europe |
| NA | North America |
| LATAM | Latin America & Caribbean |
| APAC | Asia-Pacific |
| SA | South Asia |
| MEA | Middle East & Africa |
| Other | Unmapped / rare countries |

### Hypothesis

If the predictive signal is primarily **country-specific**, then replacing `country` with `region` will lead to a measurable drop in performance, as coarse regional groupings cannot capture fine-grained distinctions.

If the signal is largely **region-driven**, then using `region` alone should achieve performance comparable to the baseline, indicating that dimensionality can be reduced without significant loss.

Adding `region` alongside the original country encoding is expected to provide little to no improvement, as it largely duplicates existing information.

### Controlled Change

Three variants are evaluated while keeping the rest of the pipeline unchanged:

| Variant | Description | Change vs baseline |
|---|---|---|
| **A - Original Only** | OHE country (baseline) - no region column | None |
| **B - Region + Original** | Keep OHE country and add `region` | +1 categorical feature (8 categories) |
| **C - Region Only** | Replace `country` with `region` (OHE, 8 categories) | Replace ~169 OHE columns with 8 |

````python
#  Country → Region mapper (ISO 3166-1 alpha-3 codes)
COUNTRY_TO_REGION = {
    # Western Europe
    "PRT":"WE","GBR":"WE","FRA":"WE","ESP":"WE","DEU":"WE","ITA":"WE",
    "IRL":"WE","NLD":"WE","BEL":"WE","CHE":"WE","AUT":"WE","SWE":"WE",
    "NOR":"WE","DNK":"WE","FIN":"WE","GRC":"WE","LUX":"WE","MCO":"WE",
    "AND":"WE","MLT":"WE","ISL":"WE","LIE":"WE","CYP":"WE","SMR":"WE","VAT":"WE",

    # Eastern Europe
    "RUS":"EE","POL":"EE","CZE":"EE","HUN":"EE","ROU":"EE","BGR":"EE",
    "SVK":"EE","SVN":"EE","HRV":"EE","SRB":"EE","UKR":"EE","BLR":"EE",
    "MDA":"EE","BIH":"EE","ALB":"EE","MKD":"EE","MNE":"EE","EST":"EE",
    "LVA":"EE","LTU":"EE","ARM":"EE","GEO":"EE","AZE":"EE","KAZ":"EE","TUR":"EE",

    # North America
    "USA":"NA","CAN":"NA","MEX":"NA",

    # Latin America
    "BRA":"LATAM","ARG":"LATAM","COL":"LATAM","CHL":"LATAM","PER":"LATAM",
    "VEN":"LATAM","ECU":"LATAM","URY":"LATAM","BOL":"LATAM","PRY":"LATAM",
    "GUY":"LATAM","SUR":"LATAM","DOM":"LATAM","CUB":"LATAM","JAM":"LATAM",
    "TTO":"LATAM","GTM":"LATAM","HND":"LATAM","CRI":"LATAM","PAN":"LATAM",
    "NIC":"LATAM","SLV":"LATAM","BLZ":"LATAM","HTI":"LATAM",

    # Asia-Pacific
    "CHN":"APAC","JPN":"APAC","KOR":"APAC","HKG":"APAC","TWN":"APAC",
    "SGP":"APAC","THA":"APAC","MYS":"APAC","IDN":"APAC","PHL":"APAC",
    "VNM":"APAC","AUS":"APAC","NZL":"APAC",

    # South Asia
    "IND":"SA","PAK":"SA","BGD":"SA","LKA":"SA","NPL":"SA","AFG":"SA",

    # Middle East & Africa
    "SAU":"MEA","ARE":"MEA","IRN":"MEA","IRQ":"MEA","ISR":"MEA",
    "EGY":"MEA","NGA":"MEA","ZAF":"MEA","KEN":"MEA","ETH":"MEA",
}

def map_region(country_code):
    return COUNTRY_TO_REGION.get(str(country_code).strip(), "Other")

# Apply mapping
X_train_e4 = X_train.copy()
X_test_e4  = X_test.copy()
X_train_e4["region"] = X_train_e4["country"].map(map_region).fillna("Other")
X_test_e4["region"]  = X_test_e4["country"].map(map_region).fillna("Other")

region_dist = X_train_e4["region"].value_counts()
print("Region distribution (training set):")
print(region_dist.to_string())
print(f"\nUnmapped → 'Other': {(X_train_e4['region'] == 'Other').sum()} rows "
      f"({(X_train_e4['region'] == 'Other').mean():.2%})")
print(f"OHE country columns (baseline): {X_train['country'].nunique()} unique values")
print(f"OHE region columns (new)      : {X_train_e4['region'].nunique()} unique values  (reduction: "
      f"{X_train['country'].nunique()} → {X_train_e4['region'].nunique()})")

#  Variant B: OHE country + region (add region to categorical)
categorical_features_e4b = categorical_features + ["region"]
preprocessor_e4b = ColumnTransformer(
    transformers=[
        ("binary", clone(binary_transformer),     binary_features),
        ("num",    clone(continuous_transformer),  continuous_features),
        ("cat",    clone(categorical_transformer), categorical_features_e4b),
        ("agent", clone(agent_transformer), ["agent"]),
    ], remainder="drop"
)
preprocessor_e4b.set_params(sparse_threshold=0)
pipe_e4b = ImbPipeline(steps=[
    ("preprocessor", preprocessor_e4b),
    ("smote",        SMOTE(random_state=42)),
    ("classifier",   RandomForestClassifier(**RF_BASE_PARAMS))
])
print("\nRunning Variant B (region + original country) ...")
scores_e4b = cross_validate(pipe_e4b, X_train_e4, y_train,
                             cv=cv_strategy, scoring=scoring_partC, n_jobs=-1)

#  Variant C: region only - drop country column
categorical_features_e4c = [f for f in categorical_features if f != "country"] + ["region"]
preprocessor_e4c = ColumnTransformer(
    transformers=[
        ("binary", clone(binary_transformer),     binary_features),
        ("num",    clone(continuous_transformer),  continuous_features),
        ("cat",    clone(categorical_transformer), categorical_features_e4c),
    ], remainder="drop"
)
preprocessor_e4c.set_params(sparse_threshold=0)
pipe_e4c = ImbPipeline(steps=[
    ("preprocessor", preprocessor_e4c),
    ("smote",        SMOTE(random_state=42)),
    ("classifier",   RandomForestClassifier(**RF_BASE_PARAMS))
])
print("Running Variant C (region only, country dropped) ...")
scores_e4c = cross_validate(pipe_e4c, X_train_e4, y_train,
                             cv=cv_strategy, scoring=scoring_partC, n_jobs=-1)

# Collect results
e4_variants = {
    "A - Original Only\n(OHE country)": {
        "f1_m": b_f1_mean,  "f1_s": b_f1_std,
        "rec_m": b_rec_mean, "prec_m": b_prec_mean, "prec_s": b_prec_std,
        "roc_m": b_roc_mean,
    },
    "B - Region + Original\n(keep country)": {
        "f1_m":   np.mean(scores_e4b["test_f1"]),
        "f1_s":   np.std(scores_e4b["test_f1"]),
        "rec_m":  np.mean(scores_e4b["test_recall"]),
        "prec_m": np.mean(scores_e4b["test_precision"]),
        "prec_s": np.std(scores_e4b["test_precision"]),
        "roc_m":  np.mean(scores_e4b["test_roc_auc"]),
    },
    "C - Region Only\n(drop country)": {
        "f1_m":   np.mean(scores_e4c["test_f1"]),
        "f1_s":   np.std(scores_e4c["test_f1"]),
        "rec_m":  np.mean(scores_e4c["test_recall"]),
        "prec_m": np.mean(scores_e4c["test_precision"]),
        "prec_s": np.std(scores_e4c["test_precision"]),
        "roc_m":  np.mean(scores_e4c["test_roc_auc"]),
    },
}

print("\nExperiment 4 - Country Region Encoding variant comparison")
print("-" * 72)
for vname, v in e4_variants.items():
    df1 = (v["f1_m"]   - b_f1_mean)   * 100
    dp  = (v["prec_m"] - b_prec_mean) * 100
    vn  = vname.replace("\n", " ")
    print(f"  {vn:<42s}  F1: {v['f1_m']:.4f} ({df1:+.2f}pp)  "
          f"Prec: {v['prec_m']:.4f} ({dp:+.2f}pp)  Rec: {v['rec_m']:.4f}")

e4_keys = list(e4_variants.keys())
e4_f1s_all   = [e4_variants[k]["f1_m"]   for k in e4_keys]
e4_f1_mean   = e4_variants[e4_keys[2]]["f1_m"]   # Variant C for ablation log
e4_f1_std    = e4_variants[e4_keys[2]]["f1_s"]
e4_prec_mean = e4_variants[e4_keys[2]]["prec_m"]
e4_prec_std  = e4_variants[e4_keys[2]]["prec_s"]
e4_rec_mean  = e4_variants[e4_keys[2]]["rec_m"]
e4_roc_mean  = e4_variants[e4_keys[2]]["roc_m"]
````

## Step 5: Summary

### What model inspection told us

The fitted champion grows trees to an **average depth of 47.4** (range 38-60, std 4.9)
with no depth constraint. The average leaf holds **26.95 samples** - 5× the configured
`min_samples_leaf=5` minimum, indicating the leaf constraint is not tightly binding.
Feature importance is top-heavy: `country_PRT` (0.096), `lead_time` (0.0901), `agent_encoded` (0.083),
`deposit_type_Non Refund` (0.076), and `total_of_special_requests` (0.075) together account for
**42%** of total Gini importance.

---

### Experiment 1 - `max_depth` Sweep: hypothesis **confirmed**

Constraining tree depth produced a **large, systematic, monotonic F1 degradation**:

| max_depth | F1 | Precision | vs Baseline |
|---|---|---|---|
| 5 | 0.7467  | 0.8438 | −8.79 pp |
| 10 | 0.7923 | 0.8156 | −4.23 pp |
| 15 | 0.8150 | 0.8267 | −1.96 pp |
| 20 | 0.8263 | 0.8408 | −0.08 pp |
| 30 | 0.8331 | 0.8504 | −0.02 pp |
| None | **0.8340** | **0.8499** | reference |

Severely limiting depth (e.g., `max_depth=5`) leads to substantial performance loss (−8.79 pp), indicating that the model requires sufficient depth to capture multi-feature interaction chains (e.g., lead time × deposit type × market segment × special requests).

Performance improves steadily as depth increases, but **gains diminish beyond `max_depth=30`**, where results are nearly identical to the unconstrained model (`None`). This indicates that while deeper trees are beneficial, **most useful interactions are already captured by depth ≈30**.

Therefore, although `max_depth=None` yields the highest F1, the improvement over `max_depth=30` is marginal (**+0.09 pp**). This suggests that `max_depth=30` may be a more efficient and robust choice, offering comparable performance with reduced model complexity and potentially lower overfitting risk.

Overall, the hypothesis that shallow trees underperform is confirmed, but the results also reveal a **performance plateau**, highlighting a trade-off between maximum performance and model simplicity.

---

### Experiment 2 - `room_type_mismatch` (3 Variants): hypothesis **partially disproved**

| Variant | F1 | Precision | Recall | vs Baseline F1 |
|---|---|---|---|---|
| A - Original only | 0.8335 | 0.8496 | 0.8180 | - |
| B - Eng + Original | 0.8374 | 0.8459 | 0.8290 | +0.28 pp |
| C - Eng Only | **0.8382** | 0.8447 | **0.8318** | **+0.36 pp** |

The hypothesis that the mismatch flag would add no new signal was **not fully supported**.
Variant C - replacing the two raw room-type categorical columns entirely with the binary
`room_type_mismatch` flag - achieves the **highest F1 and recall** across all variants,
with a +0.36 pp F1 gain and +1.18 pp recall gain over baseline. Precision trades off slightly
(−0.51 pp), but the net effect is positive.

This suggests the binary flag encodes the guest-dissatisfaction signal more cleanly than
two multi-category OHE columns, removing noise from the individual room-type distributions
and focusing the model on the directional mismatch. The raw categorical columns carry
redundant and slightly noisy representation of the same underlying signal.

---

### Experiment 3 - `min_samples_leaf` Sweep: hypothesis **overturned**

| min_samples_leaf | F1 | Precision | vs Baseline |
|---|---|---|---|
| **1** | **0.8517** | **0.8757** | **+1.82 pp** |
| 5 (baseline) | 0.8334 | 0.8500 | - |
| 10 | 0.8260 | 0.8413 | −0.74 pp |
| 20 | 0.8178 | 0.8351 | −1.56 pp |
| 50 | 0.8067 | 0.8263 | −2.67 pp |
| 100 | 0.7972 | 0.8249 | −3.62 pp |

The hypothesis that `min_samples_leaf=5` was near the performance peak was **not supported**.  
The sweep shows that `min_samples_leaf=1` achieved the highest cross-validated performance, with a **+1.82 percentage point improvement in F1** and a **+2.57 percentage point increase in Precision** relative to the baseline.

Performance declines steadily as `min_samples_leaf` increases, indicating that the baseline setting was likely **over-regularising** the model.

Even though smaller leaf sizes are typically associated with a higher risk of overfitting, **no evidence of poorer generalisation was observed in the cross-validation results**. This suggests that the dataset contains meaningful fine-grained interactions (e.g., country × deposit type × lead time) that benefit from more granular splits. However we will still treat this with caution as allowing `min_samples_leaf=1` would mean a more complex tree that overfits data.

Overall, the experiment demonstrates that `min_samples_leaf=5` was not optimal within this tested range. While `min_samples_leaf=1` provides the best performance in this case, further validation (e.g., broader tuning or hold-out testing) is recommended before adopting it as the final configuration.

## Consolidated Interpretation – What the Results Reveal About the Model

The experiments collectively reveal that the Random Forest model relies on a combination of **deep interaction structures, fine-grained feature granularity, and selective feature simplification**, rather than any single dominant factor.

First, the model's strong dependence on **tree depth** indicates that cancellation behaviour is driven by **high-order feature interactions**. The severe performance drop at low `max_depth` confirms that shallow trees cannot capture these relationships. However, the performance plateau observed beyond depth ≈30 suggests that while depth is necessary, **fully unconstrained growth is not required**, and most useful interaction patterns are already captured at moderate depth.

Second, the results highlight the importance of **feature representation quality over raw dimensionality**. The `room_type_mismatch` experiment shows that a well-designed binary feature can outperform multiple raw categorical features by **encoding the underlying signal more directly and reducing noise**. This demonstrates that the model benefits from **informative abstraction**, not just more features.

In contrast, the country-to-region encoding experiment shows that **not all dimensionality reduction is beneficial**. Replacing country with region leads to a substantial performance drop, indicating that the model depends on **fine-grained, country-level behavioural patterns**. This confirms that some high-cardinality features are **load-bearing** and should not be aggressively compressed.

Third, the `min_samples_leaf` results indicate that the model benefits from **granular partitioning of the data**. Smaller leaf sizes improve performance, suggesting that meaningful patterns exist at a fine resolution. However, this also introduces a trade-off, as smaller leaves increase model complexity and potential overfitting risk, even if this was not observed in cross-validation.

Overall, the model can be characterised as:
- **Interaction-heavy**: requires sufficient depth to model complex feature combinations  
- **Granularity-sensitive**: benefits from fine-grained splits and detailed features  
- **Selective in feature compression**: responds well to targeted feature engineering, but degrades when critical detail is lost  

These findings imply that effective improvements should focus on:
- preserving **important high-resolution features** (e.g., country-level signals),  
- introducing **meaningful engineered features** that simplify noisy inputs, and  
- balancing **model complexity and generalisation** through controlled regularisation rather than aggressive constraints.

### Step 1: Fit Champion and Generate Test Set Predictions

The champion Random Forest pipeline is refit on the full training set and used to generate:

A results DataFrame is constructed that joins the raw feature values with the actual label, predicted label, and confidence score. This allows us to filter directly for misclassified instances and rank them by how confident the model was when it was wrong.

The confusion matrix is printed to confirm the overall error counts before drilling into individual cases.

````python
#Step 1: Fit champion model and predict on test set

from sklearn.metrics import confusion_matrix

# Refit champion on full training data
champion_pipeline.fit(X_train, y_train)

# Predicted labels and confidence scores
y_pred       = champion_pipeline.predict(X_test)
y_proba      = champion_pipeline.predict_proba(X_test)[:, 1]

# Build a results frame aligned to X_test
results_df = X_test.copy().reset_index(drop=True)
results_df["actual"]          = y_test.values
results_df["predicted"]       = y_pred
results_df["confidence_cancel"] = y_proba

# Derive confidence for the predicted class
results_df["confidence"] = results_df.apply(
    lambda r: r["confidence_cancel"] if r["predicted"] == 1 else 1 - r["confidence_cancel"],
    axis=1
)

print("Confusion matrix (test set):")
cm = confusion_matrix(y_test, y_pred)
print(pd.DataFrame(cm,
                   index=["Actual: Not-Canceled", "Actual: Canceled"],
                   columns=["Pred: Not-Canceled", "Pred: Canceled"]))
````

### Feature Values

After inspecting the raw feature values of the 10 highest-confidence errors it

### Three Root Causes Identified

| Root Cause | Cases | Description |
|------------|-------|-------------|
| Zero engagement signals masked by dominant lead time | FP-1, FP-2, FP-3, FP-4 | Long lead time and Online TA + No Deposit overwhelmed all other signals; model could not distinguish genuine low-engagement bookings from risky ones |
| Segment-level base rates applied too broadly to individuals | FP-2, FP-5, FN-4 | The model learned strong priors for Online TA and Direct bookings and applied them uniformly, even when individual-level signals contradicted the segment pattern |
| Same-day booking risk not captured | FN-1, FN-3, FN-5 | Lead_time=0 is treated like any other short lead time; the model has no specific feature to flag the elevated cancellation risk of same-day bookings |
| Loyalty and segment priors suppress last-minute risk signals | FN-2, FN-4 | A returning Corporate guest (FN-2) and a Groups booking (FN-4) were treated as safe due to strong segment priors, overriding the risk from their short lead time and No Deposit |

### Proposed Fixes in Priority Order

**Fix 1 — Combine the features instead of reading them separately**
A long lead time looks risky on its own, but combining it with engagement signals forces the model to consider commitment level together with timing: (special_requests + booking_changes) / lead_time

**Fix 2 — Stop treating all Online TA and No Deposit bookings the same**
The model flags this combination as risky because it cancels often
on average. Adding a feature that captures the actual cancellation
rate per booking segment gives the model a more precise signal.

**Fix 3 — Add a cancellation rate per booking segment**
4 out of 5 False Negatives had lead_time=0. The model treats
same-day bookings as low-risk because short lead times correlate
with stays in training. Adding a binary is_same_day_booking flag
gives the model a direct signal for this specific high-risk pattern
it currently cannot see.

**Fix 4 — Lower the decision threshold**
45 high-confidence missed cancellations versus only 8 wrongly flagged
stays shows the model is biased toward predicting stays. Lowering the
threshold below 0.5 would catch more real cancellations.

### Limitations

Some failures represent patterns that are difficult to eliminate through
feature engineering alone. Same-day cancellations in particular — seen
in FN-1, FN-3, and FN-5 — may result from last-minute changes in plans
that leave no trace in the booking record. The most practical response
to these systematic False Negatives is threshold tuning in Part E rather
than adding more features.

### Part D Extension: Implementing the Proposed Fixes

The four fixes proposed in the failure analysis are implemented here as controlled experiments. Each fix is applied in isolation against the same baseline champion pipeline, and compared on F1, Recall, and Precision to determine whether the fix produces a measurable improvement.

### Baseline Reference

The baseline is the champion Random Forest pipeline from Part C — no additional features, no threshold changes. It is fit on `X_train` and evaluated on the held-out `X_test`. All fix experiments use the same train/test split for a direct comparison.

````python
# Baseline: fit champion RF on X_train, evaluate on X_test
from sklearn.base import clone

baseline_fix_pipe = ImbPipeline(steps=[
    ("preprocessor", preprocessor),
    ("smote",        SMOTE(random_state=42)),
    ("classifier",   RandomForestClassifier(
                         n_estimators=200,
                         min_samples_leaf=5,
                         n_jobs=-1,
                         random_state=42))
])

baseline_fix_pipe.fit(X_train, y_train)
base_pred  = baseline_fix_pipe.predict(X_test)

base_f1   = f1_score(y_test, base_pred, zero_division=0)
base_rec  = recall_score(y_test, base_pred)
base_prec = precision_score(y_test, base_pred, zero_division=0)

print(f"Baseline  F1={base_f1:.4f}  Recall={base_rec:.4f}  Precision={base_prec:.4f}")
````

### Fix 1: Engagement Ratio Feature

**Hypothesis:** Long lead time combined with zero engagement (no special requests, no booking changes) is the primary driver of False Positives. The model cannot distinguish a genuine long-lead booking from a cancellation risk because both look identical in the feature space. Adding a normalised engagement ratio `(total_of_special_requests + booking_changes) / max(lead_time, 1)` gives the model a signal for commitment level relative to how far ahead the booking was made.

**Implementation:** The feature is computed from existing columns after the train-test split and added to both `X_train` and `X_test`. It is treated as a continuous feature and processed by the existing numeric branch of the preprocessor.

````python
# Fix 1: Engagement ratio — (special_requests + booking_changes) / max(lead_time, 1)
X_train_f1 = X_train.copy()
X_test_f1  = X_test.copy()

X_train_f1["engagement_ratio"] = (
    (X_train_f1["total_of_special_requests"] + X_train_f1["booking_changes"])
    / X_train_f1["lead_time"].clip(lower=1)
)
X_test_f1["engagement_ratio"] = (
    (X_test_f1["total_of_special_requests"] + X_test_f1["booking_changes"])
    / X_test_f1["lead_time"].clip(lower=1)
)

continuous_f1 = continuous_features + ["engagement_ratio"]

preprocessor_f1 = ColumnTransformer(
    transformers=[
        ("binary", binary_transformer,     binary_features),
        ("num",    continuous_transformer,  continuous_f1),
        ("cat",    categorical_transformer, categorical_features),
        ("agent",  agent_transformer,       ["agent"]),
    ],
    remainder="drop"
)
preprocessor_f1.set_params(sparse_threshold=0)

pipe_f1 = ImbPipeline(steps=[
    ("preprocessor", preprocessor_f1),
    ("smote",        SMOTE(random_state=42)),
    ("classifier",   RandomForestClassifier(
                         n_estimators=200, min_samples_leaf=5,
                         n_jobs=-1, random_state=42))
])

pipe_f1.fit(X_train_f1, y_train)
pred_f1 = pipe_f1.predict(X_test_f1)

f1_f1   = f1_score(y_test, pred_f1, zero_division=0)
rec_f1  = recall_score(y_test, pred_f1)
prec_f1 = precision_score(y_test, pred_f1, zero_division=0)

print(f"Fix 1 (engagement_ratio)  F1={f1_f1:.4f}  Recall={rec_f1:.4f}  Precision={prec_f1:.4f}")
print(f"Delta vs baseline         F1={f1_f1-base_f1:+.4f}  Recall={rec_f1-base_rec:+.4f}  Precision={prec_f1-base_prec:+.4f}")
````

### Fix 2: Joint Target Encoding of (market_segment, deposit_type)

**Hypothesis:** The model applies a blanket high-risk prior to all Online TA + No Deposit bookings, even when individual-level signals suggest the guest will stay. The root cause is that `market_segment` and `deposit_type` are encoded independently — the model never sees their combined cancellation rate directly. Creating a joint feature `segment_deposit` that concatenates both values, then target-encoding it, gives the model a precise cancellation rate for each specific combination.

**Implementation:** A new string column `segment_deposit` is derived by concatenating `market_segment` and `deposit_type`. It is added to the categorical branch and target-encoded using the same smoothed `TargetEncoder` already used for `agent`.

````python
# Fix 2: Joint target encoding of (market_segment, deposit_type)
X_train_f2 = X_train.copy()
X_test_f2  = X_test.copy()

X_train_f2["segment_deposit"] = (
    X_train_f2["market_segment"] + "_" + X_train_f2["deposit_type"]
)
X_test_f2["segment_deposit"] = (
    X_test_f2["market_segment"] + "_" + X_test_f2["deposit_type"]
)

segment_deposit_transformer = SkPipeline(steps=[
    ("imputer",    SimpleImputer(strategy="most_frequent")),
    ("target_enc", TargetEncoder(smooth="auto"))
])

preprocessor_f2 = ColumnTransformer(
    transformers=[
        ("binary",  binary_transformer,            binary_features),
        ("num",     continuous_transformer,        continuous_features),
        ("cat",     categorical_transformer,       categorical_features),
        ("agent",   agent_transformer,             ["agent"]),
        ("seg_dep", segment_deposit_transformer,   ["segment_deposit"]),
    ],
    remainder="drop"
)
preprocessor_f2.set_params(sparse_threshold=0)

pipe_f2 = ImbPipeline(steps=[
    ("preprocessor", preprocessor_f2),
    ("smote",        SMOTE(random_state=42)),
    ("classifier",   RandomForestClassifier(
                         n_estimators=200, min_samples_leaf=5,
                         n_jobs=-1, random_state=42))
])

pipe_f2.fit(X_train_f2, y_train)
pred_f2 = pipe_f2.predict(X_test_f2)

f1_f2   = f1_score(y_test, pred_f2, zero_division=0)
rec_f2  = recall_score(y_test, pred_f2)
prec_f2 = precision_score(y_test, pred_f2, zero_division=0)

print(f"Fix 2 (segment_deposit enc)  F1={f1_f2:.4f}  Recall={rec_f2:.4f}  Precision={prec_f2:.4f}")
print(f"Delta vs baseline            F1={f1_f2-base_f1:+.4f}  Recall={rec_f2-base_rec:+.4f}  Precision={prec_f2-base_prec:+.4f}")
````

### Fix 3: is_same_day_booking Flag + Lead Time Buckets

**Hypothesis:** 4 out of 5 False Negatives had `lead_time = 0` or `lead_time <= 15`. The model treats same-day and very last-minute bookings as low-risk because short lead times correlate with stays in training. Two features address this:

- `is_same_day_booking` — binary flag set to 1 when `lead_time == 0`, giving the model a direct signal for the specific elevated-risk same-day pattern
- `lead_time_bucket` — ordinal bucketing of lead time into `[0, 1–7, 8–30, 31–90, 90+]` to allow the model to learn non-linear risk patterns per range rather than treating raw lead time as a continuous gradient

Both features are leakage-safe as they are derived purely from `lead_time` with no target information.

````python
# Fix 3: is_same_day_booking flag + lead_time_bucket
X_train_f3 = X_train.copy()
X_test_f3  = X_test.copy()

X_train_f3["is_same_day_booking"] = (X_train_f3["lead_time"] == 0).astype(int)
X_test_f3["is_same_day_booking"]  = (X_test_f3["lead_time"]  == 0).astype(int)

bins   = [-1, 0, 7, 30, 90, float("inf")]
labels = ["same_day", "1_to_7", "8_to_30", "31_to_90", "90_plus"]
X_train_f3["lead_time_bucket"] = pd.cut(
    X_train_f3["lead_time"], bins=bins, labels=labels
).astype(str)
X_test_f3["lead_time_bucket"] = pd.cut(
    X_test_f3["lead_time"], bins=bins, labels=labels
).astype(str)

binary_f3      = binary_features + ["is_same_day_booking"]
categorical_f3 = categorical_features + ["lead_time_bucket"]

preprocessor_f3 = ColumnTransformer(
    transformers=[
        ("binary", binary_transformer,     binary_f3),
        ("num",    continuous_transformer,  continuous_features),
        ("cat",    categorical_transformer, categorical_f3),
        ("agent",  agent_transformer,       ["agent"]),
    ],
    remainder="drop"
)
preprocessor_f3.set_params(sparse_threshold=0)

pipe_f3 = ImbPipeline(steps=[
    ("preprocessor", preprocessor_f3),
    ("smote",        SMOTE(random_state=42)),
    ("classifier",   RandomForestClassifier(
                         n_estimators=200, min_samples_leaf=5,
                         n_jobs=-1, random_state=42))
])

pipe_f3.fit(X_train_f3, y_train)
pred_f3 = pipe_f3.predict(X_test_f3)

f1_f3   = f1_score(y_test, pred_f3, zero_division=0)
rec_f3  = recall_score(y_test, pred_f3)
prec_f3 = precision_score(y_test, pred_f3, zero_division=0)

print(f"Fix 3 (same_day + lt_bucket)  F1={f1_f3:.4f}  Recall={rec_f3:.4f}  Precision={prec_f3:.4f}")
print(f"Delta vs baseline             F1={f1_f3-base_f1:+.4f}  Recall={rec_f3-base_rec:+.4f}  Precision={prec_f3-base_prec:+.4f}")
````

### Fix 4: Threshold Lowering

**Hypothesis:** The model produces far more high-confidence False Negatives than False Positives, indicating it is systematically biased toward predicting stays. Lowering the decision threshold below the default 0.5 increases sensitivity, catching more real cancellations at the cost of some additional false alarms. The optimal threshold is chosen by maximising Recall while keeping Precision above a minimum acceptable floor of 0.75.

**Implementation:** The champion pipeline is refit on the full training set. Thresholds from 0.30 to 0.50 are swept and the impact on FP/FN counts is reported. This fix operates on the existing champion pipeline — no new features are added.

````python
# Fix 4: Threshold lowering on the champion pipeline
# Refit champion on full training set
champion_pipeline.fit(X_train, y_train)
y_proba_fix4 = champion_pipeline.predict_proba(X_test)[:, 1]

print(f"{'Threshold':>10}  {'Precision':>10}  {'Recall':>10}  {'F1':>10}  {'FP':>6}  {'FN':>6}")
print("-" * 62)

fix4_rows = []
for t in np.arange(0.30, 0.51, 0.05):
    preds = (y_proba_fix4 >= t).astype(int)
    prec  = precision_score(y_test, preds, zero_division=0)
    rec   = recall_score(y_test, preds)
    f1    = f1_score(y_test, preds, zero_division=0)
    fp    = int(((preds == 1) & (y_test == 0)).sum())
    fn    = int(((preds == 0) & (y_test == 1)).sum())
    fix4_rows.append({"threshold": round(t, 2), "precision": prec,
                      "recall": rec, "f1": f1, "FP": fp, "FN": fn})
    print(f"{t:>10.2f}  {prec:>10.4f}  {rec:>10.4f}  {f1:>10.4f}  {fp:>6}  {fn:>6}")

fix4_df = pd.DataFrame(fix4_rows)

# Recommended threshold: best recall with precision >= 0.75
valid = fix4_df[fix4_df["precision"] >= 0.75]
if not valid.empty:
    best_fix4 = valid.loc[valid["recall"].idxmax()]
    print(f"\nRecommended threshold (max recall, precision >= 0.75): {best_fix4['threshold']:.2f}")
    print(f"  Precision={best_fix4['precision']:.4f}  Recall={best_fix4['recall']:.4f}  F1={best_fix4['f1']:.4f}")
    print(f"  FP={int(best_fix4['FP'])}  FN={int(best_fix4['FN'])}")
````

## Part E: Business Decision

### Common business strategy
One strategy many hotels use is to oversell rooms which are flagged as probable cancellations. The objective is to keep the overall occupancy as close to 100% as possible. Since there is a chance that the hotel will receive more occupants and available rooms, it is not ideal having to move or reject incoming occupants as more expenses would be incurred in the process. The **walk** cost is the cost of  arranging contigency stay for overbooked occupants.

### Receiver Operating Characteristic(ROC) Curve
The ROC curve is very close to the top-left of the graph, almost hugging the corner. This reveals that the model is very high discriminative power, abling to separate cancellations from non-cancellations across all possible thresholds. An AUC of 0.95 would means there is a **95%** chance that the model will give the cancelled booking a higher predicted probability than the non-cancelled one. Therefore, the model is reliable enough to support threshold tuning.

````python
# ── 1. Sweep thresholds ──────────────────────────────────────────────────────
thresholds = np.arange(0.10, 0.91, 0.01)

precisions, recalls, f1s = [], [], []
for t in thresholds:
    preds = (y_proba >= t).astype(int)
    precisions.append(precision_score(y_test, preds, zero_division=0))
    recalls.append(recall_score(y_test, preds, zero_division=0))
    f1s.append(f1_score(y_test, preds, zero_division=0))

best_f1_idx = int(np.argmax(f1s))
best_thresh = thresholds[best_f1_idx]
print(f"Best F1 threshold : {best_thresh:.2f}  (F1 = {f1s[best_f1_idx]:.4f})")

# ── 2. Threshold sensitivity plot ────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Part E — Threshold Sensitivity Analysis (Champion Random Forest)",
             fontsize=13, fontweight="bold")

# Left: Precision / Recall / F1 vs threshold
ax = axes[0]
ax.plot(thresholds, precisions, color="#378ADD", linewidth=2.0, label="Precision")
ax.plot(thresholds, recalls,   color="#1D9E75", linewidth=2.0, label="Recall")
ax.plot(thresholds, f1s,       color="#D85A30", linewidth=2.5, label="F1 Score")
ax.axvline(0.50,        color="gray",    linestyle="--", linewidth=1.2, label="Default (0.50)")
ax.axvline(best_thresh, color="#D85A30", linestyle=":",  linewidth=1.5,
           label=f"Best F1 ({best_thresh:.2f})")
for t_mark, label_mark in [(0.35, "0.35\n(aggressive)"), (0.65, "0.65\n(conservative)")]:
    ax.axvline(t_mark, color="#888780", linestyle="-.", linewidth=1.0, alpha=0.7)
    ax.text(t_mark + 0.005, 0.44, label_mark, fontsize=7.5, color="#888780", va="bottom")

ax.set_xlabel("Decision Threshold", fontsize=10)
ax.set_ylabel("Score", fontsize=10)
ax.set_title("Precision / Recall / F1 vs Threshold", fontsize=11, fontweight="bold")
ax.set_xlim(0.10, 0.90)
ax.set_ylim(0.40, 1.02)
ax.legend(fontsize=9, loc="lower left")
ax.grid(axis="both", linestyle="--", alpha=0.4)

# Right: Precision-Recall curve with operating points marked
ax2 = axes[1]
ax2.plot(recalls, precisions, color="#378ADD", linewidth=2.0, label="PR curve")

key_thresholds = [0.35, 0.50, 0.65]
key_colors     = ["#1D9E75", "#888780", "#D85A30"]
for kt, kc in zip(key_thresholds, key_colors):
    idx = int(round((kt - 0.10) / 0.01))
    ax2.scatter(recalls[idx], precisions[idx], color=kc, s=80, zorder=5,
                edgecolors="white", linewidths=1.2,
                label=f"t={kt:.2f}  P={precisions[idx]:.3f}  R={recalls[idx]:.3f}")

ax2.set_xlabel("Recall", fontsize=10)
ax2.set_ylabel("Precision", fontsize=10)
ax2.set_title("Precision–Recall Curve with Key Operating Points", fontsize=11, fontweight="bold")
ax2.legend(fontsize=8.5, loc="lower left")
ax2.grid(linestyle="--", alpha=0.4)

plt.tight_layout()
plt.savefig("partE_threshold_sensitivity.png", dpi=150, bbox_inches="tight")
plt.show()

# ── 3. Estimated Error Impact Table ──────────────────────────────────────────
# Assumptions (adjust to your hotel's actual figures):
ADR_SGD       = 200   # average daily rate per room (SGD)
WALK_COST_SGD = 300   # estimated cost per walked guest if overbooking fires (SGD)

impact_rows = []
for t in [0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60, 0.65, 0.70]:
    preds = (y_proba >= t).astype(int)
    tp = int(((preds == 1) & (y_test == 1)).sum())
    fp = int(((preds == 1) & (y_test == 0)).sum())
    fn = int(((preds == 0) & (y_test == 1)).sum())
    tn = int(((preds == 0) & (y_test == 0)).sum())
    prec = precision_score(y_test, preds, zero_division=0)
    rec  = recall_score(y_test, preds, zero_division=0)
    f1   = f1_score(y_test, preds, zero_division=0)
    rev_at_risk = fn * ADR_SGD        # FN → empty rooms → lost revenue
    walk_cost   = fp * WALK_COST_SGD  # FP → false alarms → walk cost if overbooking
    total_cost  = rev_at_risk + walk_cost   # new computation

    impact_rows.append({
        "Threshold":              f"{t:.2f}",
        "Precision":              round(prec, 4),
        "Recall":                 round(rec,  4),
        "F1 Score":               round(f1,   4),
        "False Positives (FP)":   fp,
        "False Negatives (FN)":   fn,
        "Revenue at Risk (SGD)":  f"${rev_at_risk:,}",
        "Walk Cost Risk (SGD)":   f"${walk_cost:,}",
        "Total Risk (SGD)": f"${total_cost:,}",
    })

impact_df = pd.DataFrame(impact_rows)

def highlight_default(row):
    color = "background-color: #343e85" if row["Threshold"] == "0.50" else ""
    return [color] * len(row)

print("=" * 80)
print("Estimated Error Impact Table")
print("Assumptions: ADR = SGD $200 per room | Walk cost = SGD $300 per false-alarm guest")
print("Purple row = default 0.50 threshold (current champion operating point)")
print("=" * 80)
display(impact_df.style.apply(highlight_default, axis=1))

print("\nBusiness interpretation:")
print(f"  Each FN (missed cancel) → 1 empty room not re-sold → ~SGD ${ADR_SGD} lost revenue")
print(f"  Each FP (false alarm)   → potential guest walk if overbooking → ~SGD ${WALK_COST_SGD} cost")
print(f"  Scenario A (minimise empty rooms)  → lower threshold toward 0.35 reduces FN count")
print(f"  Scenario B (overbooking strategy)  → raise threshold toward 0.65 reduces FP / walk risk")
````

### Threshold Sensitivity Analysis
The champion Random Forest outputs a **probability** via `predict_proba()`, not a hard label. The default `predict()` applies a 0.5 cutoff, any booking with `P(cancel) >= 0.5` is flagged as a cancellation.

The cutoff is sweep from 0.10 to 0.90 and record Precision, Recall, and F1 at each point, then display an **Estimated Error Impact Table** at key thresholds to quantify the business cost of False Negatives (missed cancellations → empty rooms) and False Positives (false alarms → potential guest walks).

### Score vs Threshold
From the Score vs Threshold graph, we can see that the best F1 score reside at 0.45 threshold, suggesting that the model's mathematically optimal balance point is slighty below the default 0.50 threshold. The F1 curve is relatively flat between 0.35 and 0.55. Since the score does not change drastically in that region, the business will have the flexibility to shift the threshold left or right for business reasons without paying a heavy performance penalty. Beyond 0.55 and below 0.35, the tradeoffs become severe. Precision collapses rapidly below 0.35, and Recall collapses rapidly above 0.55. These are the hard boundaries outside which the model becomes operationally unreliable, which is not recommended.

### Presision Recall(PR) Curve
The gap between t=0.35 and t=0.50 on recall (0.927 vs 0.819) is much larger than the gap between t=0.50 and t=0.65 on precision (0.848 vs 0.921). Lowering the threshold buys a lot more recall than raising it for precision. This asymmetry directly favours the aggressive strategy if the business's primary concern is empty rooms.

## Business Decision
The statistically optimal threshold (t=0.45, best F1 = 0.837) and the economically optimal threshold (t=0.55, lowest total cost = SGD $689,100) diverge because false positives and false negatives carry asymmetric business costs. Deploying at t=0.55 reduces total estimated risk by SGD 77,800 compared to the best F1 threshold, and SGD 22,300 compared to the default t=0.50. It's worth noting that these cost metrics are hypothetical. If the walk costs were lowered to SGD 150, the cost-minimising threshold would shift back leftward. Hence, the threshold itself should be treated as a dynamic parameter, not a fixed number. A human-in-the-loop adjustment layer on top of the model's output is ideal for high cancellation risk (for e.g peak seasons, economic uncertainty, post-COVID patterns).
