```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline as SkPipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, RobustScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, make_scorer
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier

from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.over_sampling import SMOTE

from category_encoders import TargetEncoder

if not os.path.exists("hotel_bookings.csv"):
    raise FileNotFoundError("Please place hotel_bookings.csv in the same folder as this notebook.")

df_raw = pd.read_csv("hotel_bookings.csv")
print("Raw shape:", df_raw.shape)
df_raw.head()
```

## Structural Cleaning Before Pipeline Construction

A few preprocessing steps are applied **before** pipeline construction because they are deterministic structural corrections rather than learned transformations.

These include:

1. removing leakage variables
2. dropping the `company` column due to extreme missingness
3. removing logically invalid bookings with no adults, children, or babies
4. standardising `meal = "Undefined"` into `"None"`

All remaining learned preprocessing operations — imputation, encoding, scaling, and class balancing — are then handled strictly inside the pipeline.


```python
df = df_raw.copy()

TARGET_COL = "is_canceled"

LEAKAGE_COLS = [
    "reservation_status",
    "reservation_status_date"
]

DROP_COLS = [
    "company"
]

# Remove leakage columns and structurally unsuitable columns
# errors="ignore" keeps the notebook robust even if a column is absent
df = df.drop(columns=LEAKAGE_COLS + DROP_COLS, errors="ignore")

# Standardise a known ambiguous meal label
df["meal"] = df["meal"].replace("Undefined", "None")

# Remove logically invalid bookings: no adults, no children, and no babies
before_rows = len(df)
df = df[~((df["adults"] == 0) & (df["children"].fillna(0) == 0) & (df["babies"] == 0))]
after_rows = len(df)

print(f"Rows removed due to invalid booking composition: {before_rows - after_rows}")
print("Shape after structural cleaning:", df.shape)
df.head()
```

## Feature–Target Separation

The target variable remains **`is_canceled`** where:

- `1` = cancelled booking
- `0` = not cancelled booking

At this point the feature matrix is intentionally kept in its **raw mixed-type form**, because the transformation work will now be handled by the pipeline rather than manually on the full dataset.


```python
y = df[TARGET_COL]
X = df.drop(columns=[TARGET_COL])

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Target distribution:")
print(y.value_counts(normalize=True).rename("proportion"))
```

## Train–Test Split

The dataset is divided into training and test sets using an **80/20 split** with stratification on the target variable.

This preserves class distribution in both subsets while ensuring the **test set remains untouched** by fitting steps such as imputation, target encoding, scaling, and SMOTE.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("Target distribution (train):")
print(y_train.value_counts(normalize=True))

print("Target distribution (test):")
print(y_test.value_counts(normalize=True))
```

## Feature Grouping for Pipeline Engineering

To support different preprocessing strategies, the predictors are divided into:

- **numeric features**
- **low-cardinality categorical features**
- **high-cardinality categorical features**

This is important because not all categorical variables should be treated the same way.

- low-cardinality categorical features are suitable for **one-hot encoding**
- high-cardinality features are better handled with **target encoding** to avoid excessive sparsity

A threshold of **more than 10 unique categories** is used to define high-cardinality categorical variables.


```python
numeric_features = X_train.select_dtypes(include=["number"]).columns.tolist()
categorical_features = X_train.select_dtypes(exclude=["number"]).columns.tolist()

high_cardinality_features = [
    col for col in categorical_features
    if X_train[col].nunique(dropna=True) > 10
]

low_cardinality_features = [
    col for col in categorical_features
    if col not in high_cardinality_features
]

print("Numeric features:", numeric_features)
print("Low-cardinality categorical features:", low_cardinality_features)
print("High-cardinality categorical features:", high_cardinality_features)

print("Counts")
print("Numeric:", len(numeric_features))
print("Low-cardinality categorical:", len(low_cardinality_features))
print("High-cardinality categorical:", len(high_cardinality_features))
```

## Preprocessing Design

The preprocessing pipeline applies different transformations to different feature groups.

### 1. Numeric Features
- median imputation
- RobustScaler

This is suitable because many numeric variables in the hotel booking dataset are skewed and contain outliers.

### 2. Low-Cardinality Categorical Features
- most-frequent imputation
- OneHotEncoder(handle_unknown="ignore")

This preserves nominal category information and avoids errors from unseen categories in validation or test data.

### 3. High-Cardinality Categorical Features
- constant imputation using `"Missing"`
- TargetEncoder

This avoids the dimensional explosion that would happen if variables such as `country` or `agent` were one-hot encoded.

### 4. Class Imbalance Handling
- SMOTE is applied **after preprocessing** and **inside the pipeline**
- this ensures synthetic samples are generated only from the transformed training data


```python
# Numeric preprocessing
numeric_transformer = SkPipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", RobustScaler())
])

# Low-cardinality categorical preprocessing
low_cardinality_transformer = SkPipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

# High-cardinality categorical preprocessing
high_cardinality_transformer = SkPipeline(steps=[
    ("imputer", SimpleImputer(strategy="constant", fill_value="Missing")),
    ("target", TargetEncoder())
])

# ColumnTransformer combines all preprocessing branches
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("low_cat", low_cardinality_transformer, low_cardinality_features),
        ("high_cat", high_cardinality_transformer, high_cardinality_features)
    ],
    remainder="drop"
)

# Stage 2 leakage-safe pipeline
stage2_pipeline = ImbPipeline(steps=[
    ("preprocessor", preprocessor),
    ("smote", SMOTE(random_state=42)),
    ("classifier", LogisticRegression(max_iter=2000, random_state=42))
])

stage2_pipeline
```

# Part B: Champion Model Selection

# Candidate Model Pipeline

Each model is wrapped inside the same **ImbPipeline** structure from Part A. The pipeline runs three steps in sequence for every fold of cross-validation:

1. **Preprocessor**: applies imputation, encoding, and scaling to the raw features
2. **SMOTE**: generates synthetic minority class samples to address class imbalance, but only on the training fold so that validation data is never contaminated
3. **Classifier**: the model that learns from the processed training data and makes predictions

Keeping all three steps inside one pipeline prevents data leakage and ensures every model is evaluated under identical conditions.

```python
# Build the 3 candidate pipelines using the same preprocessor from Part A
candidate_pipelines = {

    "Logistic Regression": ImbPipeline(steps=[
        ("preprocessor", preprocessor),
        ("smote",        SMOTE(random_state=42)),
        ("classifier",   LogisticRegression(
                             C=1.0,
                             max_iter=2000,
                             random_state=42))
    ]),

    "Random Forest": ImbPipeline(steps=[
        ("preprocessor", preprocessor),
        ("smote",        SMOTE(random_state=42)),
        ("classifier",   RandomForestClassifier(
                             n_estimators=200,
                             min_samples_leaf=5,
                             n_jobs=-1,
                             random_state=42))
    ]),

    "Hist Gradient Boosting": ImbPipeline(steps=[
        ("preprocessor", preprocessor),
        ("smote",        SMOTE(random_state=42)),
        ("classifier",   HistGradientBoostingClassifier(
                             max_iter=200,
                             learning_rate=0.05,
                             max_leaf_nodes=31,
                             random_state=42))
    ]),

}
```

# 5-Fold Stratified Cross-Validation

**How it works:**


The training data is divided into 5 equal parts called folds. In each round, 4 folds are used to train the model and the remaining 1 fold is used to test it. This process repeats 5 times, rotating which fold is used for testing each time. The final reported score is the average across all 5 rounds, along with the standard deviation to show how consistent the model was.

**How it supports fair model selection:**


Using 5 folds instead of a single train/test split means the model is tested on 5 different slices of data. This reduces the risk that a lucky or unlucky split skews the result. The standard deviation also reveals whether a model performs consistently or whether its score swings widely depending on what data it sees — a stable model is more trustworthy in production.


**Stratified** means each fold is constructed so that the proportion of cancelled vs not-cancelled bookings stays the same as in the full dataset. This prevents a fold from accidentally containing too many or too few cancellations, which would distort the evaluation.


```python
# Set up 5-fold stratified cross-validation
# Stratified means each fold keeps the same class ratio as the full dataset
cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Define the metrics to collect across all folds
scoring = {
    "f1":        make_scorer(f1_score, zero_division=0),
    "recall":    "recall",
    "precision": "precision",
    "roc_auc":   "roc_auc",
}

# Run cross-validation for each model and store the results
cv_results = {}

for name, pipe in candidate_pipelines.items():
    print(f"Running cross-validation for: {name} ...")
    scores = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=cv_strategy,
        scoring=scoring,
        return_train_score=False,
    )
    cv_results[name] = scores
    print(f"  F1 Score : {scores['test_f1'].mean():.4f} ± {scores['test_f1'].std():.4f}")
    print(f"  Recall   : {scores['test_recall'].mean():.4f} ± {scores['test_recall'].std():.4f}")
    print(f"  Precision: {scores['test_precision'].mean():.4f} ± {scores['test_precision'].std():.4f}")
    print(f"  ROC-AUC  : {scores['test_roc_auc'].mean():.4f} ± {scores['test_roc_auc'].std():.4f}")
    print()
```

## Cross-Validation Results Summary Table

The summary table below presents the **mean and standard deviation** of each metric across all 5 folds.

- **Mean** reflects the average performance of the model across 5 different test slices
- **Std** reflects how stable that performance is — a lower std means the model generalises consistently regardless of which data it is tested on

The table is sorted by **F1 Score (mean)** from highest to lowest, as F1 is the primary metric used for Champion selection.

```python
rows = []
for name, scores in cv_results.items():
    rows.append({
        "Model"            : name,
        "F1 (mean)"        : round(scores["test_f1"].mean(),        4),
        "F1 (std)"         : round(scores["test_f1"].std(),         4),
        "Recall (mean)"    : round(scores["test_recall"].mean(),    4),
        "Recall (std)"     : round(scores["test_recall"].std(),     4),
        "Precision (mean)" : round(scores["test_precision"].mean(), 4),
        "Precision (std)"  : round(scores["test_precision"].std(),  4),
        "ROC-AUC (mean)"   : round(scores["test_roc_auc"].mean(),   4),
        "ROC-AUC (std)"    : round(scores["test_roc_auc"].std(),    4),
    })

summary_df = (pd.DataFrame(rows)
                .sort_values("F1 (mean)", ascending=False)
                .reset_index(drop=True))

display(summary_df)
```

# Part C: Controlled Ablations and Tuning

## Controlled Ablation Setup – Random Forest Champion Model

This section performs a **controlled ablation study** on the selected **Champion Random Forest** model. The goal is to test whether small, isolated changes to the model configuration improve generalization performance.

### Step 1: Recover the Original Champion Configuration
The champion pipeline is first refit on the full training set so that the trained Random Forest can be inspected directly.

The inspection shows that the original champion model has:
- **200 trees**
- **`min_samples_leaf = 5`**
- **no explicit `max_depth` constraint** (`max_depth=None`)

Because `max_depth` was not fixed, the trees were allowed to grow until other stopping conditions were reached. The fitted forest shows:
- average tree depth of approximately **37**
- minimum depth of **30**

This suggests that the original model learns **very deep trees**, which motivates testing whether restricting complexity can improve generalization.

## Rationale for the Four Controlled Experiments

Each experiment changes **only one factor** while keeping the rest of the pipeline unchanged. This ensures that any performance difference can be attributed to that single modification.

### 4. Experiment 3 – Class Weighting

### 5. Experiment 4 – Stronger Leaf Regularization
**Change:** `min_samples_leaf: 5 → 10`

**Rationale:**

## Evaluation Procedure

For each experiment, the same modelling pipeline is used:

- **preprocessor**
- **SMOTE**
- **Random Forest classifier**

Performance is evaluated using **5-fold cross-validation** with the following metrics:

- **F1**
- **Recall**
- **Precision**
- **ROC-AUC**

For each metric, both the **mean** and **standard deviation** are recorded. This allows the experiments to be compared not only on average performance, but also on **stability across folds**.


```python
results = []

scoring = ["f1", "recall", "precision", "roc_auc"]


for name, config in RF_EXPERIMENTS.items():

    rf_model = RandomForestClassifier(**config["params"])

    model = ImbPipeline(steps=[
        ("preprocessor", preprocessor),
        ("smote", SMOTE(random_state=42)),
        ("classifier", rf_model)
    ])

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=5,
        scoring=scoring,
        n_jobs=-1,
        error_score="raise"
    )

    results.append({
        "Experiment": name,
        "Hypothesis": config["description"],

        "F1 Mean": np.mean(scores["test_f1"]),
        "F1 Std": np.std(scores["test_f1"]),

        "Recall Mean": np.mean(scores["test_recall"]),
        "Recall Std": np.std(scores["test_recall"]),

        "Precision Mean": np.mean(scores["test_precision"]),
        "Precision Std": np.std(scores["test_precision"]),

        "ROC-AUC Mean": np.mean(scores["test_roc_auc"]),
        "ROC-AUC Std": np.std(scores["test_roc_auc"])
    })

ablation_results = pd.DataFrame(results)
ablation_results
```

### Step 1: Fit Champion and Generate Test Set Predictions

The champion Random Forest pipeline is refit on the full training set and used to generate:
- **Predicted class labels** (`y_pred`) for each test instance
- **Predicted probability of cancellation** (`y_proba`) — the model's confidence score

A results DataFrame is constructed that joins the raw feature values with the actual label, predicted label, and confidence score. This allows us to filter directly for misclassified instances and rank them by how confident the model was when it was wrong.

The confusion matrix is printed to confirm the overall error counts before drilling into individual cases.

```python
#Step 1: Fit champion model and predict on test set

from sklearn.metrics import confusion_matrix

# Refit champion on full training data
champion_pipeline.fit(X_train, y_train)

# Predicted labels and confidence scores
y_pred       = champion_pipeline.predict(X_test)
y_proba      = champion_pipeline.predict_proba(X_test)[:, 1]

# Build a results frame aligned to X_test
results_df = X_test.copy().reset_index(drop=True)
results_df["actual"]          = y_test.values
results_df["predicted"]       = y_pred
results_df["confidence_cancel"] = y_proba

# Derive confidence for the predicted class
results_df["confidence"] = results_df.apply(
    lambda r: r["confidence_cancel"] if r["predicted"] == 1 else 1 - r["confidence_cancel"],
    axis=1
)

print("Confusion matrix (test set):")
cm = confusion_matrix(y_test, y_pred)
print(pd.DataFrame(cm,
                   index=["Actual: Not-Canceled", "Actual: Canceled"],
                   columns=["Pred: Not-Canceled", "Pred: Canceled"]))
```

### Step 2: Extract High-Confidence Misclassifications

Two error types are extracted from the test set:

- **False Positives (FP):** The model predicted the booking would be *cancelled*, but the guest actually *stayed*. These would be costly if the hotel acts on the prediction (e.g., reselling the room prematurely).
- **False Negatives (FN):** The model predicted the booking would *not be cancelled*, but the guest actually *cancelled*. These would be costly if the hotel fails to manage inventory or overbooking.

For each error type, the **top 5 instances with the highest model confidence** are extracted. High-confidence errors are used because they reveal cases where the model was not just slightly off but also systematically misled by a pattern in the feature values.


```python
#Step 2: Extract high-confidence misclassifications

# False Positives: predicted cancel (1), actually not canceled (0)
fp_df = (results_df[(results_df["predicted"] == 1) & (results_df["actual"] == 0)]
         .sort_values("confidence", ascending=False)
         .head(5))

# False Negatives: predicted not-canceled (0), actually canceled (1)
fn_df = (results_df[(results_df["predicted"] == 0) & (results_df["actual"] == 1)]
         .sort_values("confidence", ascending=False)
         .head(5))

KEY_COLS = [
    "actual", "predicted", "confidence",
    "lead_time", "deposit_type", "previous_cancellations",
    "booking_changes", "total_of_special_requests",
    "market_segment", "customer_type",
    "reserved_room_type", "assigned_room_type",
    "stays_in_week_nights", "stays_in_weekend_nights",
    "adr", "is_repeated_guest"
]

print("═" * 70)
print("HIGH-CONFIDENCE FALSE POSITIVES  (predicted: Canceled | actual: Not-Canceled)")
print("═" * 70)
display(fp_df[KEY_COLS].reset_index(drop=True))

print("\n")
print("═" * 70)
print("HIGH-CONFIDENCE FALSE NEGATIVES  (predicted: Not-Canceled | actual: Canceled)")
print("═" * 70)
display(fn_df[KEY_COLS].reset_index(drop=True))
```

### Summary of Findings for Part D

The test set confusion matrix shows 1,395 False Positives and 1,492 False Negatives out of 23,842 test instances. While these numbers appear small relative to the total, the row-level inspection reveals that these errors are not random where they are traceable to specific feature patterns that the model consistently misreads.

In step 5, the confidence distribution shows:
- **8 high-confidence False Positives** (confidence ≥ 0.9) — the model was extremely certain a guest would cancel, but they stayed
- **45 high-confidence False Negatives** (confidence ≥ 0.9) — the model was extremely certain a guest would stay, but they cancelled

The model is better at spotting guests who stay then guests who cancel, this would be a huge mistake for the hotel as a cancellation that the model did not see would mean an empty room with no time to sell it to potential customers reducing hotel revenue

###Feature Values

After inspecting the raw feature values of the 10 highest-confidence errors it
showed clear patterns in both error types.

For **False Positives**, all 5 instances shared the same combination:
Online TA or Groups bookings with No Deposit, zero booking changes,
zero or one special requests, and a first-time guest profile. The model
predicted cancellation because this combination resembles the typical
high-risk profile in the training data. However, these guests stayed therefore
the model had no way to distinguish a low engagement booking from a risky one.

For **False Negatives**, the failed instances showed the opposite problem.
All 5 False Negatives had near-zero lead times (0-15 days), No Deposit, and zero prior cancellations. The model treated same-day and last-minute bookings as highly committed because short lead times correlate with stays in training. However these guests cancelled anyway — a pattern the model cannot detect without a specific feature for same-day booking risk

In general, the model failed not because the features were uninformative,
but because it treated each feature independently rather than considering
how combinations of features interact.

### Three Root Causes Identified

| Root Cause | Cases | Description |
|------------|-------|-------------|
| Zero engagement signals masked by dominant lead time | FP-1, FP-2, FP-3, FP-4 | Long lead time and Online TA + No Deposit overwhelmed all other signals; model could not distinguish genuine low-engagement bookings from risky ones |
| Segment-level base rates applied too broadly to individuals | FP-2, FP-5, FN-4 | The model learned strong priors for Online TA and Direct bookings and applied them uniformly, even when individual-level signals contradicted the segment pattern |
| Same-day booking risk not captured | FN-1, FN-3, FN-5 | Lead_time=0 is treated like any other short lead time; the model has no specific feature to flag the elevated cancellation risk of same-day bookings |
| Loyalty and segment priors suppress last-minute risk signals | FN-2, FN-4 | A returning Corporate guest (FN-2) and a Groups booking (FN-4) were treated as safe due to strong segment priors, overriding the risk from their short lead time and No Deposit |

### Proposed Fixes in Priority Order

**Fix 1 — Combine the features instead of reading them separately**
A long lead time looks risky on its own, but combining it with engagement signals forces the model to consider commitment level together with timing: (special_requests + booking_changes) / lead_time

**Fix 2 — Stop treating all Online TA and No Deposit bookings the same**
The model flags this combination as risky because it cancels often
on average. Adding a feature that captures the actual cancellation
rate per booking segment gives the model a more precise signal.

**Fix 3 — Add a cancellation rate per booking segment**
4 out of 5 False Negatives had lead_time=0. The model treats
same-day bookings as low-risk because short lead times correlate
with stays in training. Adding a binary is_same_day_booking flag
gives the model a direct signal for this specific high-risk pattern
it currently cannot see.

**Fix 4 — Lower the decision threshold**
45 high-confidence missed cancellations versus only 8 wrongly flagged
stays shows the model is biased toward predicting stays. Lowering the
threshold below 0.5 would catch more real cancellations.

### Limitations

Some failures represent patterns that are difficult to eliminate through
feature engineering alone. Same-day cancellations in particular — seen
in FN-1, FN-3, and FN-5 — may result from last-minute changes in plans
that leave no trace in the booking record. The most practical response
to these systematic False Negatives is threshold tuning in Part E rather
than adding more features.
